# Ridge Regression


The objective of this notebook is to implement Ridge Regression using Scikit-learn.

Ridge Regression is an extension of Linear Regression that introduces L2 Regularization to reduce overfitting by penalizing large coefficient values.

In this notebook, we will:

- Understand the problem of overfitting.
- Learn the concept of regularization.
- Train a Ridge Regression model.
- Study the effect of the alpha parameter.
- Predict house prices.
- Evaluate model performance.
- Compare Ridge Regression with Linear Regression.

```python
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

from sklearn.linear_model import Ridge         # import lib

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
project_root = Path.cwd().parent              # Data transformed from data transformation

train_path = project_root / "artifacts" / "data_transformation" / "train.npy"
test_path = project_root / "artifacts" / "data_transformation" / "test.npy"

train_arr = np.load(train_path)
test_arr = np.load(test_path)
```

```python
X_train = train_arr[:, :-1]
y_train = train_arr[:, -1]   # The transformed arrays contain both the input features and the target variable.

X_test = test_arr[:, :-1]
y_test = test_arr[:, -1]
```

```python
print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

# Create the Ridge Regression Model

The Ridge Regression model is created using Scikit-learn.

The alpha parameter controls the amount of L2 regularization applied during training.

 Smaller alpha → Less regularization.
 Larger alpha → Stronger regularization.

```python
model = Ridge(alpha=1.0)        # Create the Ridge Regression Model
                                  
model.fit(X_train, y_train)
```

```python
print(model)
```

```python
model.fit(X_train, y_train)

print("Ridge Regression model trained successfully!")
```

```python
print("Intercept (Bias):")      # Learned Parameters here parameter are bias and weight
print(model.intercept_)         # only single bias is present Intercept = intercept for learned parameter value 

print("\nCoefficients (Weights):") # Ridge Regression applies L2 regularization, which shrinks the coefficient values to improve generalization
print(model.coef_)        #coefficient value is nothing but weights . coeeficient = coef for model coeff checking 

# in ridge shrinks the slope /coeff/weight value i.e., features but never its '0'
```

```python
y_pred = model.predict(X_test)      # pedict price i.e., target variable by using X_test y is the target variable 

print("Predicted House Prices:")
print(y_pred[:5])
```

```python
comparison = np.column_stack((y_test, y_pred))    # Compare Actual vs Predicted

print("Actual Price\tPredicted Price")
print(comparison[:10])
```

# Model Evaluation

The Ridge Regression model is evaluated using:

- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R² Score

These metrics help assess prediction accuracy and compare Ridge Regression with other regression models.

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)
```

```python
print(f"Mean Absolute Error (MAE): {mae:.2f}")                    # Display results
                                          
print(f"Mean Squared Error (MSE): {mse:.2f}")

print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")

print(f"R² Score: {r2:.4f}")
```
