# Linear Regression using Scikit-learn


The objective of this notebook is to implement Linear Regression using Scikit-learn's built-in `LinearRegression` model.

In the previous notebook, Linear Regression was implemented manually using Gradient Descent. Every mathematical step, including prediction, cost calculation, gradient computation, and parameter updates, was coded from scratch.

In this notebook, Scikit-learn automates the optimization process, allowing us to train the model using a single `fit()` method.

The notebook covers:

 Loading the transformed dataset.
 Training the Linear Regression model.
 Understanding model coefficients and intercept.
 Predicting house prices.
 Evaluating model performance.
 Visualizing prediction results.

```python
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

from sklearn.linear_model import LinearRegression     # import lib 

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
project_root = Path.cwd().parent

train_path = project_root / "artifacts" / "data_transformation" / "train.npy"
test_path = project_root / "artifacts" / "data_transformation" / "test.npy"

train_arr = np.load(train_path)
test_arr = np.load(test_path)     
# load transformed data : The dataset has already been: Cleaned, Encoded, Split into training and testing sets 
```

```python
X_train = train_arr[:, :-1]
y_train = train_arr[:, -1]   # The transformed arrays contain both the input features and the target variable.

X_test = test_arr[:, :-1]
y_test = test_arr[:, -1]
```

```python
print(X_train.shape)
print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

```python
model = LinearRegression()
                               # Create the Linear Regression Mode
model.fit(X_train, y_train)
```

```python
print("Model trained successfully!")
```

```python
print("Intercept (Bias):")           # Inspect the Learned Parameters i.e., bias and weight
print(model.intercept_)

print("\nCoefficients (Weights):")
print(model.coef_)
```

```python
y_pred = model.predict(X_test)    # Make Predictions on Test Data
```

```python
print("Predicted House Prices:")
print(y_pred[:5])
```

```python
comparison = np.column_stack((y_test, y_pred))

print("Actual Price\tPredicted Price")   # Compare Actual vs Predicted Values
print(comparison[:10])
```

# Model Evaluation

The trained model is evaluated using the following metrics:

 Mean Absolute Error (MAE)
 Mean Squared Error (MSE)
 Root Mean Squared Error (RMSE)
 R² Score

These metrics help measure the prediction accuracy and compare different machine learning models.

```python
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)                    #checking all values for better prediction 

r2 = r2_score(y_test, y_pred)
```

```python
print(f"Mean Absolute Error (MAE): {mae:.2f}")

print(f"Mean Squared Error (MSE): {mse:.2f}")

print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")   #Display the Results

print(f"R² Score: {r2:.4f}")
```
